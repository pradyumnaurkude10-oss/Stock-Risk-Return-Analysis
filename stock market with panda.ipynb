```python
print("Mean of BEML" , np.round(beml1["gain"].mean(),6))
print("Standard deviation of BEML" , np.round(beml1["gain"].std(),6))
 
```

```python
print("Mean of GLAXO" , np.round(glaxo1["gain"].mean(),6))
print("Standard deviation od GLAXO" , np.round(glaxo1["gain"].std(),6))
```

```python
beml_mean = beml1["gain"].mean()
beml_std = beml1["gain"].std()

glaxo_mean = glaxo1["gain"].mean()
glaxo_std = glaxo1["gain"].std()
```

```python
stats.norm.cdf(-0.02,beml_mean,beml_std) # here is 22% chance of incurring 2% loss or higher in beml
```

```python
stats.norm.cdf(-0.02,glaxo_mean,glaxo_std) # here is 6% chance of incurring 2% loss or higher in glaxo
```

```python
1- stats.norm.cdf(0.02,beml_mean,beml_std)
```

```python
1- stats.norm.cdf(0.02,glaxo_mean,glaxo_std)
```

```python
stats.norm.ppf(0.975)
```

```python
np.round(stats.norm.ppf(0.975),2) 
```

```python
stats.norm.ppf(0.95)
```

```python
np.round(stats.norm.ppf(0.95),2)
```

```python
#confidence intervals

beml_ci = stats.norm.interval(0.95,beml_mean,beml_std)
```

```python
np.round(beml_ci,4)
```

```python
glaxo_ci = stats.norm.interval(0.95,glaxo_mean,glaxo_std)
glaxo_ci
```
